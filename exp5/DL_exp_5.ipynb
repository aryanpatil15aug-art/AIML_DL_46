```python
column_name=[
    'age','sex','cp','trestbps','chol',
    'fbs','restecg','thalach', 'exang',
    'oldpeak','slope','ca','thal'
]
df=pd.read_csv("heart_disease.csv")
```

```python
df.isnull().sum()
```

```python
X=df.drop('target',axis=1)
Y=df['target']

scaler=StandardScaler()
X=scaler.fit_transform(X)

X_train,X_test,Y_train,Y_test=train_test_split(
    X,Y,
    test_size=0.2,
    random_state=42
)
```

```python
X_train=torch.FloatTensor(X_train)
Y_train=torch.FloatTensor(Y_train.values).view(-1,1)
X_test=torch.FloatTensor(X_test)
Y_test=torch.FloatTensor(Y_test.values).view(-1,1)
```

```python
print(X_train.shape)
X_train=X_train[:90]
Y_train=Y_train[:90]
```

```python
model=nn.Sequential(
    nn.Linear(13,64),
    nn.ReLU(),
    nn.Linear(64,32),
    nn.ReLU(),
    nn.Linear(32,16),
    nn.ReLU(),
    nn.Linear(16,1),
    nn.Sigmoid()
)
```

```python
criterion=nn.BCELoss()
optimizer=optim.Adam(model.parameters(),lr=0.001)
for epoch in range(500):
    output=model(X_train)
    train_loss=criterion(output, Y_train)
        
    optimizer.zero_grad()
    train_loss.backward()
    optimizer.step()
with torch.no_grad():
    output=model(X_train)
    train_prediction=(output>=0.5).float()
    test_output = model(X_test)
    test_loss = criterion(test_output, Y_test)
    output=model(X_test)
    test_prediction=(output>=0.5).float()
    
train_accuracy = (train_prediction == Y_train).float().mean() * 100
test_accuracy = (test_prediction == Y_test).float().mean() * 100

print("Train Accuracy:",train_accuracy.item())
print(f"Train Loss:{train_loss.item():.10f}")

print("Test Accuracy:",test_accuracy.item())
print(f"Test Loss:{test_loss.item():.10f}")
```
