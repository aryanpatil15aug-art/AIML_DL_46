```python
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
```

```python
X,Y = make_moons(n_samples=200,noise=0.2,random_state=42)
X=torch.FloatTensor(X)
Y=torch.FloatTensor(Y).view(-1,1)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(
    X,Y,test_size=0.2,random_state=42
)
```

#### Part A: OPTIMIZERS

```python
def train_model(optim_name="Adam"):
    model=nn.Sequential(
        nn.Linear(2,16),
        nn.ReLU(),
        nn.Linear(16,1),
        nn.Sigmoid()
    )
    criterion=nn.BCELoss()
    if optim_name=="SGD":
        optimizer=optim.SGD(model.parameters(),lr=0.01)
    elif optim_name=="Momentum":
        optimizer=optim.SGD(model.parameters(),lr=0.01,momentum=0.9)
    elif optim_name=="RMSProp":
        optimizer=optim.RMSprop(model.parameters(),lr=0.01)
    elif optim_name=="Adam":
        optimizer=optim.Adam(model.parameters(),lr=0.01)
    
    epochs=500
    for epoch in range(epochs):
        output=model(X_train)
        loss=criterion(output,Y_train)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
    with torch.no_grad():
        output=model(X_test)
        prediction=(output>=0.5).float()
    test_accuracy = (prediction == Y_test).float().mean() * 100
    return test_accuracy.item(),loss.item()
```

```python
optimizers=["SGD","Momentum","RMSProp","Adam"]
print("Optimizer\t Loss\t Accuracy")
for opt in optimizers:
    acc,loss=train_model(optim_name=opt)
    print(f"{opt}\t\t{loss:.2f}\t\t{acc:.2f}%")
```

#### Part B: ACTIVATION FUNCTIONS

```python
def train_model(activation_name):
    if activation_name=="Sigmoid":
        activation=nn.Sigmoid()
    elif activation_name=="Tanh":
        activation=nn.Tanh()
    elif activation_name=="ReLU":
        activation=nn.ReLU()
    elif activation_name=="LeakyReLU":
        activation=nn.LeakyReLU()
        
    model=nn.Sequential(
        nn.Linear(2,16),
        activation,
        nn.Linear(16,1),
        nn.Sigmoid()
    )
    criterion=nn.BCELoss()
    optimizer=optim.Adam(model.parameters(),lr=0.1)
    for epoch in range(1000):
        output=model(X_train)
        loss=criterion(output, Y_train)
        
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
    with torch.no_grad():
        output=model(X_test)
        prediction=(output>=0.5).float()
    test_accuracy = (prediction == Y_test).float().mean() * 100
    return test_accuracy.item(),loss.item()
```

```python
activation_func=["Sigmoid","Tanh","ReLU","LeakyReLU"]
print("Activation Func\tLoss\t Accuracy")
for act in activation_func:
    acc,loss=train_model(activation_name=act)
    print(f"{act}\t\t{loss:.2f}\t\t{acc:.2f}%")
```
