#### Part A: Implement AND Logic Gates using a Perceptron

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Input and Output
X = torch.tensor([[0.,0.],
                  [0.,1.],
                  [1.,0.],
                  [1.,1.]])
Y = torch.tensor([[0.],
                  [0.],
                  [0.],
                  [1.]])
```

```python
# Initialize weights and bias
W = torch.randn(2, 1, requires_grad=True)
b = torch.randn(1, requires_grad=True)
learning_rate = 0.1
```

```python
# Training
for epoch in range(1000):

    # Forward propagation
    z=X @ W + b
    output = torch.sigmoid(z)

    # Binary Cross Entropy Loss
    # loss = -(Y * torch.log(output) +
    #          (1 - Y) * torch.log(1 - output)).mean()
    criterion = nn.BCELoss()
    loss = criterion(output,Y)

    # Backward propagation computes gradients
    loss.backward()

    # Update weights
    with torch.no_grad():
        W -= learning_rate * W.grad
        b -= learning_rate * b.grad

        W.grad.zero_()
        b.grad.zero_()
```

```python
# Testing
with torch.no_grad():
    # prediction = (torch.sigmoid(X @ W + b) >= 0.5).float()         #coverts boolean to float
    prediction = (output >= 0.5).float()         #coverts boolean to float
```

```python
print("Predictions")
print(prediction)

accuracy = (prediction == Y).float().mean() * 100
print("\nAccuracy =", accuracy.item(), "%")

print("\nFinal Weights")
print(W)

print("\nFinal Bias")
print(b)
```

#### Part B: Solving AND Gate Using NN Library

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Input and Output
X = torch.tensor([[0.,0.],
                  [0.,1.],
                  [1.,0.],
                  [1.,1.]])
Y = torch.tensor([[0.],
                  [0.],
                  [0.],
                  [1.]])
```

```python
# Initialize weights and bias
W = torch.randn(2, 1, requires_grad=True)
b = torch.randn(1, requires_grad=True)
learning_rate = 0.1
```

```python
model=nn.Sequential(
    nn.Linear(2,1),         # create 2 input 1 output
    nn.Sigmoid()
)
```

```python
criterion = nn.BCELoss()
optimizer = optim.SGD(model.parameters(),lr=0.1)
for epoch in range(1000):
    output = model(X)
    loss=criterion(output,Y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

with torch.no_grad():
    prediction=(output>=0.5).float()

print("Prediction")
print(prediction)
```

```python
print("\nFinal Weights")
print(W)

print("\nFinal Bias")
print(b)
```
